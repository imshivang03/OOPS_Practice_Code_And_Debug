```python
## Implementing Encapsulation

class Bank():

    def __init__(self, holder_name: str, balance: str):
        self.holder_name: str = holder_name
        # Private Attribute
        self.__balance: int = balance

    # Private Method
    def __is_server_live(self):
        return True

    def get_balance(self) -> None:
        print(f"Hey {self.holder_name}, Your bank balance is {self.__balance}")

    def set_balance(self, bal: int) -> None:
        self.__balance: int = bal

    def deposit(self, amount:int) -> None:
        if self.__is_server_live():
            self.__balance += amount
            print(f"Amount deposited: {amount} and your updated balance is {self.__balance}")
        else:
            print("Server is Down, try after sometime")

    def withdraw(self, amount:int) -> None:
        if self.__is_server_live():
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Amount withdrawn: {amount} and current balance is {self.__balance}")
            else:
                print(f"You don't have enough balance to withdraw, your current balance is {self.__balance}")
        else:
            print("Server is Down, try after sometime")


acc=Bank("Shivang Shahi", 2000)
acc.get_balance()
acc.deposit(5000)
acc.withdraw(1500)
acc.set_balance(12000)
acc.get_balance()
```
