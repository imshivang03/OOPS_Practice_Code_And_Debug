```python
class Animal:

    def __init__(self, name:str, age:int):
        self.name=name
        self.age=age

    def eat(self):
        print("I am eating")

    def sleep(self):
        print("I am sleeping")

class Dog(Animal):
    def bark(self):
        print("I am barking")

    def display(self):
        print(f"Dog's name is {self.name} and the age is {self.age}")

d=Dog("Boxer", 5)
d.display()
```

```python
class Animal:

    def __init__(self, name:str, age:int) -> None:
        print("This is animal's init")
        self.name=name
        self.age=age

    def eat(self):
        print("I am eating")

    def sleep(self):
        print("I am sleeping")


class Dog(Animal):

    def __init__(self, name:str, age:int, breed:str) -> None:
        print("This is dog's init")
        super().__init__(name, age)
        self.breed=breed

    def bark(self):
        print("I am barking")

    def display(self):
        print(f"Dog's name: {self.name}, age: {self.age} and breed: {self.breed}")

d=Dog("Boxer", 5, "German Shepherd")
d.display()
print("---------------------")
d1=Dog("Maxx", 2, "Labrador")
d1.display()
```
