```python
class Student:

    def __init__(self, name: str, age: int, gender: str) -> None:
        # Attributes
        self.name=name
        self.age=age
        self.gender=gender

    def display(self) -> None:
        print(f"Your name is {self.name}, your age is {self.age} and your gender is {self.gender}")
    
    def get_age(self) -> int:
        print(f"age: {self.age}")
        return self.age

s1=Student("Shivang Shahi", 28, "male")
s1.display()
s1.get_age()
print(s1.get_age())

s2=Student("Yashaswi Shahi", 25, "male")
s2.display()
s2.get_age()
print(s2.get_age())
```

```python
class Movie():
    
    # Attributes
    def __init__(self, movie_name, total_seats, ticket_price):
        self.movie_name=movie_name
        self.total_seats=total_seats
        self.ticket_price=ticket_price
        self.booked_seats=0

    def show_status(self,):
        print(f"movie name is {self.movie_name}")
        print(f"available seats: {self.total_seats}")
        print(f"ticket price: {self.ticket_price}")
        if self.booked_seats==0:
            print(f"No bookings done yet.")
        else:
            print(f"Seats booked so far: {self.booked_seats}")

    def book_tickets(self, num_of_tickets):
        if num_of_tickets > self.total_seats:
            print("Sorry, not enough seats available")
        else:
            print("-----------Booking started-----------")
            print()
            print(f"Available seats: {self.total_seats}")
            self.total_seats-=num_of_tickets
            self.booked_seats+=num_of_tickets
            print(f"Seats booked: {self.booked_seats}")
            print(f"Seats left: {self.total_seats}")
            print(f"Ticket Price: {self.ticket_price*self.booked_seats}")
            print()
            print("-----------Booking completed-----------")


m1=Movie("Dhoom", 100, 200)

m1.show_status()
print()
print()

m1.book_tickets(15)
print()
print()

m1.show_status()
```
